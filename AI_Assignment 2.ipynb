```python
team_members = int(input("Number of team members: "))
months = int(input("Number of project months: "))
monthly_salary = float(input("Monthly salary per team member: "))
cloud_cost = float(input("Total cloud cost for the project: "))

salary_cost = team_members * months * monthly_salary
subtotal = salary_cost + cloud_cost
vat = subtotal * 0.14
total = subtotal + vat

print(f"Salary cost: {salary_cost:,.2f}")
print(f"Subtotal: {subtotal:,.2f}")
print(f"VAT (14%): {vat:,.2f}")
print(f"Total project cost: {total:,.2f}")
```
